# Explainability analyses

Runs one explainability analysis of the article on the stored checkpoints of the inversions (run `FWI.ipynb`,
`Ablations.ipynb` or `run_all.sh` first) and draws the corresponding figure of the article.

**Keys**
- `model`: `'over'` or `'marm'`
- `analysis` (module of `xai/`, output in `results/xai/<analysis>/<model>_linear/`):
  - `'landscape'`: misfit along a time shift and along the line from the initial to the true model (Figs 5 and S3)
  - `'adjoint'`: adjoint sources, integrated gradients and alignment of the update (Figs 6, 7, S4 and S5)
  - `'occlusion'`: effect of replacing early or later arrivals on the misfit and on the update (Figs 7 and S5)
  - `'identity'`: departure of the SiameseFWI network from the identity (Table S5)
  - `'fixed_data'`: adjoint sources of the networks of each epoch on fixed data (Figs 9 and S7)
  - `'cka'`: linear CKA and weight changes of the SiameseFit encoder (Figs 8 and S6)
  - `'depth'`: SNR per depth band of the final models (Figs 3 and 10)
  - `'training_stats'`: data-independent output of the trained SiameseFit networks (Table S5)
  - `'cross_misfit'` (Marmousi2): misfits at the final models of other inversions (Table 5)
  - `'updates'`: updates along the inversion and single-shot gradients (Table S6)

Requirements: the runs of seeds 0 to 2 (seeds 0 to 4 for `identity`, `cka` and `training_stats`) of the misfits used by
each analysis; see the docstring of each module. Most analyses take a few minutes on a GPU.


```python
model = 'over'          # 'over' or 'marm'
analysis = 'landscape'  # see the list above
```

```python
%run -m xai.$analysis --model $model
```

Figure of the article built from the analysis (all figures and tables: `Figures.ipynb`).

```python
from IPython.display import Image, display
from utils import figures

figure = {'landscape': 'landscape', 'adjoint': 'adjoint_initial', 'occlusion': 'alignment', 'fixed_data': 'training_fixed',
          'cka': 'sfit_epochs', 'depth': 'performance'}.get(analysis)
name = {'over': 'overthrust', 'marm': 'marmousi2'}[model]
if figure:
    figures.set_style()
    figures.FIGURES[figure]([name])
    display(Image(figures.FIG_DIR + '/fig_%s_%s.png' % (figure, name)))
```
